# ****Week 1 - Task 2: Data Cleaning & Visualization****

# Step 1: Import Required Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

#Loading dataset
df=pd.read_csv("train.csv")

#Selecting plot style
sns.set_style("whitegrid")

#checking missing values
df.isnull().sum()
```

# Step 3: Handle Missing Values

```python
#filling age missing values
df["Age"].fillna(df["Age"].median(), inplace=True)

#We have used median because age has some extra values and is less affected by outliers
```

```python
df["Embarked"].fillna(df["Embarked"].mode()[0], inplace=True)
#filling embarked missing values
```

```python
#Cabin column contain 75% missing values
df.drop(columns=["Cabin"], inplace=True)
```

```python
#Verifying missing values
df.isnull().sum()
```

# Step 4: Add Markdown Justification

## Handling Missing Values

- The Age column contained missing values, which were filled using the median age because the median is less affected by extreme values and outliers.
- The Embarked column had only a few missing values, so they were filled with the most frequent category (mode).
- The Cabin column had a large number of missing values (more than 75%), making it unreliable for analysis. Therefore, it was removed from the dataset.

# Step 10: Analyze Survival Factors

```python
#Lets compare survival by factors

plt.figure(figsize=(8,5))

sns.countplot(
    x="Sex",
    hue="Survived",
    data=df

)
plt.title("Survival by Gender")
plt.show()

#chart usually showing that most females survived than men
```
